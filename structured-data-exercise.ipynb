```python
"""
Objective: Practice using prompt engineering and stop sequencing to 
structure Claude's output to be more readable and usable
"""

# Identify changes in source files and reload
%load_ext autoreload
%autoreload 2

from dotenv import load_dotenv
from anthropic import Anthropic
from structured_data_exercise import StructuredDataExercise
```

```python
# Access the API key
load_dotenv()
```

```python
# Access the Anthropic API
client = Anthropic()
# Specify the model Claude will use
model = "claude-sonnet-5"

"""
Access:
- Conversation history related to structurizing Claude output data
- Functions to store user inputs
- Functions to store Claude responses
"""
structuredDataExercise = StructuredDataExercise(client, model)
```

```python
# System prompt to tell Claude how to approach the problem
system_prompt = "You are an expert around all things AWS. Return only AWS CLI commands. No comments. No explanation."

# Prompt engineering rules to structure the output data
initial_prompt = """
Generate exactly three different sample AWS CLI commands.

Rules:
- Return only plain text.
- Do not use markdown.
- Do not use code fences.
- Put each command on its own line.
- Do not include comments.
- Do not include explanations.
- After the third command, write END_OF_COMMANDS.
"""

# Tells Claude when to stop generating
stop_sequence = "END_OF_COMMANDS"
```

```python
# Apply the stop sequence to Claude's prompts
structuredDataExercise.stop_sequences.append("END_OF_COMMANDS")

# Start the conversation with a user message
structuredDataExercise.storeUserInput(initial_prompt)

# Apply the system prompt to Claude's prompts
claudeResponse = structuredDataExercise.askClaude(system_prompt)
```

```python
# Store Claude's response
structuredDataExercise.storeClaudeResponse(claudeResponse)
```
