```python
"""
Objective: Make a chat bot using the three helper functions
from Multi-turn conversations
"""

# Import API key and Anthropic API
from dotenv import load_dotenv
from anthropic import Anthropic

# Import chat bot helper functions
from chat_bot_functions import ChatBot
```

```python
# Access the API key
load_dotenv()

# Access the Anthropic API
client = Anthropic()

# Specify the model Claude will use
model = "claude-sonnet-5"

# Access helper functions
chat_bot = ChatBot(client, model)
```

```python
# Prompt the user to enter some input using the built-in "input" function
initial_prompt = "Where do you want to go on vacation?"

print(initial_prompt)

# Store the user's response
user_input = input()

print(user_input)
```

```python
# Add the user's response to the conversation history
extended_user_input = f"I want to go vacation to {user_input}. What can you tell me about this place in one sentence?"
chat_bot.storeUserInput(extended_user_input)

# Get Claude's response to the user's input with the conversation history
claude_response = chat_bot.askClaude()

# Add Claude's response to the conversation history
chat_bot.storeClaudeResponse(claude_response)

print(claude_response)
```

```python
"""
Objective: Demonstrate that Claude can retain context to future questions by leveraging the conversation history
"""
# End case to stop the conversation
continue_conversation = "yes"
extended_follow_up_question = f"Ask your next questions about {user_input}"
ask_another_question = "Would you like to ask another question? yes / no"

while(continue_conversation == "yes"):
    print(extended_follow_up_question)
    user_input = input()

    truncated_user_input = f"{user_input}, answer in one sentence"
    chat_bot.storeUserInput(truncated_user_input)

    claude_response = chat_bot.askClaude()
    chat_bot.storeClaudeResponse(claude_response)

    print(claude_response)

    print(ask_another_question)

    continue_conversation = input()
```
